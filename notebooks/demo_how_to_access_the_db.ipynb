```python
import numpy as np
import pandas as pd
import os
import dotenv
import psycopg
from sqlalchemy import create_engine
dotenv.load_dotenv()
POSTGRES_PASSWORD = os.getenv('POSTGRES_PASSWORD')
```

```python
# SQL Alchemy engine

dbms = 'postgresql'
package = 'psycopg'
user = 'postgres'
password = POSTGRES_PASSWORD
host = 'localhost'
port = '5432'
db = 'contrans'

engine = create_engine(f'{dbms}+{package}://{user}:{password}@{host}:{port}/{db}')
```

```python
myquery = '''
SELECT *
FROM expenditures
WHERE expenditure_amount > 5000
'''

pd.read_sql_query(myquery, con=engine)
```
